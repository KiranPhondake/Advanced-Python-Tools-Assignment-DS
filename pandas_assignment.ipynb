```python
# importing pands library and set common alias 

import pandas as pd
```

# 1. Create a DataFrame with the following data:

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [24, 27, 22, 32, 29],
    'Department': ['HR', 'Finance', 'IT', 'Marketing', 'HR'],
    'Salary': [45000, 54000, 50000, 62000, 47000]
}
```

```python
df =  pd.DataFrame(data)
```

```python
df
```

* pd.DataFrame(data) → converts the dictionary into a structured table (DataFrame).

 # 1a. Print the first five rows of the DataFrame.

```python
df.head()
```

* head() displays the first 5 rows by default.

* It helps us quickly verify that the DataFrame was created correctly.

# 1b. Get the summary statistics of the 'Age' and 'Salary' columns.

```python
stats = round(df[['Age', 'Salary']].describe(),2)
stats
```

* describe() provides summary statistics (count, mean, min, max, quartiles).

* Selecting columns using double brackets [['Age', 'Salary']] limits output to those two

# 1c. Calculate the average salary of employees in the 'HR' department.

```python
hr_avg_sal = df[df['Department'] == 'HR']['Salary'].mean()
print("Average HR Salary:", hr_avg_sal)
```

* df['Department'] == 'HR' → filters rows where Department is HR.

* [‘Salary’].mean() selects only the Salary column and using mean() it will return average salary.

# 2. Add a new column, 'Bonus', which is 10% of the salary.

```python
df['Bonus'] = df['Salary'] * 0.10
```

```python
df
```

* Creates a new column called Bonus.

* Multiplies each employee’s salary by 10% (0.10).

# 3. Filter the DataFrame to show employees aged between 25 and 30.

```python
filtered_df = df[(df['Age']>=25) & (df['Age']<=30)]
```

```python
filtered_df 
```

* Uses Boolean operators to select rows where age is between 25 and 30.

* The & symbol means AND (both conditions must be true).

# 4. Group the data by 'Department' and calculate the average salary for each department.

```python
avg_salary_by_dept  = df.groupby('Department')['Salary'].mean()
```

```python
avg_salary_by_dept
```

* groupby('Department') groups rows by department name.

* ['Salary'].mean() calculates the average salary per department.

# 5. Sort the DataFrame by 'Salary' in ascending order and save the result to a new CSV file.

```python
sorted_df = df.sort_values(by='Salary', ascending=True)
```

```python
sorted_df
```

```python
sorted_df.to_csv('sorted_employees.csv', index=False)
```

* sort_values(by='Salary', ascending=True) sorts rows in ascending order of Salary column.

* to_csv('sorted_employees.csv', index=False) saves the result to a CSV file without row numbers.
