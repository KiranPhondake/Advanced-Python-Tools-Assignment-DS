# 1. Create a 1D NumPy array with integers from 1 to 20

```python
arr_1d = np.arange(1, 21)
print("1D Array:\n", arr_1d)
```

* np.arange(1, 21) creates an array of integers from 1 to 20 (end value 21 is exclusive).

### a. Calculate the sum, mean, median, and standard deviation

```python
sum_arr = np.sum(arr_1d)
mean_arr = np.mean(arr_1d)
median_arr = np.median(arr_1d)
std_arr = np.std(arr_1d)
```

```python
print("\nArray Statistics:")
print(f"Sum: {sum_arr}")
print(f"Mean: {mean_arr}")
print(f"Median: {median_arr}")
print(f"Standard Deviation: {std_arr}")
```

* np.sum() → Adds up all elements

* np.mean() → Calculates the arithmetic mean

* np.median() → Finds the median value

* np.std() → Computes the standard deviation

### b. Find the indices of elements greater than 10

```python
indices_gt_10 = np.where(arr_1d > 10)
```

```python
print("\nIndices of elements greater than 10:", indices_gt_10)
```

* np.where(condition) returns the indices of array elements where the condition is True.

* Here, it finds all positions where values are greater than 10. 

```python
print("Elements greater than 10:", arr_1d[indices_gt_10])
```

* arr_1d[indices_gt_10] retrieves those actual elements.

# 2. Create 2D NumPy array of shape 4X4 with numbers ranging from 1 to 16.

```python
arr_2d = np.arange(1, 17).reshape(4,4)
```

* np.arange(1, 17).reshape(4,4) → Creates values from 1 to 16 and Converts the 1D array into a 4×4 (2D) matrix.

### a. Print the array.  

```python
print("2D Array:\n", arr_2d)
```

### b. Transpose of the Array

```python
transpose_arr = arr_2d.T
print("\nTranspose of Array:\n", transpose_arr)
```

* .T gives the transpose, i.e., swaps rows and columns.

### c. Calculate the row-wise and column-wise sums of the array.

```python
row_sums = np.sum(arr_2d, axis=1)
col_sums = np.sum(arr_2d, axis=0)
print("\nRow-wise sums:", row_sums)
print("Column-wise sums:", col_sums)
```

* axis=1 → Operates horizontally (row-wise).
* axis=0 → Operates vertically (column-wise).
* NumPy uses the axis parameter to define the direction of the operation.

# 3. Create two 3 X 3 arrays filled with random integers between 1 and 20.

```python
arr_3dA = np.random.randint(1, 21, size=(3, 3))
arr_3dB = np.random.randint(1, 21, size=(3, 3))
```

```python
print("\nArray A:\n", arr_3dA)
print("Array B:\n", arr_3dB)
```

* np.random.randint(1, 21, size=(3, 3)) creates a 3×3 array filled with random integers between 1 and 20.

* Each run it will produce different random values.

### a. Perform element-wise addition, subtraction, and multiplication.

```python
add_result = arr_3dA + arr_3dB
sub_result = arr_3dA - arr_3dB
mul_result = arr_3dA * arr_3dB

print("\nElement-wise Addition:\n", add_result)
print("\nElement-wise Subtraction:\n", sub_result)
print("\nElement-wise Multiplication:\n", mul_result)
```

*  NumPy supports vectorized operations, meaning arithmetic is done element-by-element automatically:

* arr_a + arr_b adds each corresponding element.

* arr_a - arr_b subtracts them.

* arr_a * arr_b multiplies them element-wise (not matrix multiplication).

### b. Compute the dot product of the two arrays.

```python
dot_product = np.dot(arr_3dA, arr_3dB)
print("\nDot Product of A and B:\n", dot_product)
```

* np.dot() performs matrix multiplication (sum of products of rows and columns).

* This is different from element-wise multiplication — it’s a linear algebra operation.

# Reshape a 1D array of size 12 into a 3 X 4 2D array and slice the first two rows and last two columns.

```python
arr_reshaped = np.arange(1, 13).reshape(3, 4)
print("\nReshaped 3x4 Array:\n", arr_reshaped)
```

* np.arange(1, 13) .reshape(3, 4)  creates values from 1 to 12 and converts  a 3×4 matrix.

```python
sliced_arr = arr_reshaped[:2, -2:]
print("\nSliced Array (first 2 rows, last 2 columns):\n", sliced_arr)
```

* [:2, -2:] uses slicing:

* :2 → Selects first 2 rows (index 0 and 1).

* -2: → Selects last 2 columns.
