```python
#importing numpy, pandas and matplotlib libraries 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 1a. Create a simple line plot for the following data:

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 25, 30, 50]
```

```python
plt.plot(x, y)
plt.show()
```

* plt.plot(x, y) creates a line plot with x values on the X-axis and y values on the Y-axis.

* plt.show() displays the plot.

## 1b. Customize the plot by adding a title, axis labels, and a grid.  

```python
plt.plot(x, y, color='Purple',marker='o', linestyle='-', linewidth=2)
plt.title("Simple Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.grid(True)
plt.show()
```

* color, marker,linestyle and linewidth customize the line’s appearance.

* plt.title() adds a title.

* plt.xlabel() and plt.ylabel() label the axes.

* plt.grid(True) displays a grid for better readability.

## 2 a. Create a bar graph to represent the marks scored by students in a subject:  

```python
students = ['John', 'Jane', 'Alice', 'Bob']
marks = [75, 85, 60, 90]
```

```python
plt.bar(students,marks)
plt.show()
```

## 2b. Customize the colors and add a title.  

```python
plt.bar(students, marks, color=['blue', 'green', 'purple', 'orange'])
plt.title("Marks Scored by Students")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.ylim(0, 100)  # set y-axis limit
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

* color adds individual colors for each bar.

* plt.ylim(0, 100) ensures the marks are visible from 0–100.

* plt.grid(axis='y') adds grid lines along the Y-axis only.

## 3 a. Create a pie chart to represent the percentage distribution of a company’s revenue from different regions:  
  

```python
regions = ['North America', 'Europe', 'Asia', 'Others']
revenue = [45, 25, 20, 10]
  
```

```python
#Create a pie chart with region names as labels

plt.pie(revenue, labels=regions, autopct='%1.0f%%')
plt.title("Company Revenue Distribution by Region")
plt.show()
```

* labels=regions shows region names.

* autopct='%1.1f%%' displays percentages with one decimal place.

 ## 3 b. Highlight the region with the highest revenue.  

```python
explode = [0.1,0, 0, 0]  # "explode" (pull out) the first highest slice 
plt.pie(revenue, labels=regions, autopct='%1.0f%%', explode=explode, shadow=True, startangle=90)
plt.title("Highlight Region with the Highest Revenue")
plt.show()
```

* explode separates one hoghest slice from the pie for emphasis.

* shadow=True adds a shadow effect.

* startangle=90 starts the pie from the top for better visualization.

## 4. Generate a histogram to show the frequency distribution of randomly generated integers between 1 and 100 (sample size = 1000).  


```python
data = np.random.randint(1, 101, 1000)
```

```python
data
```

```python
plt.hist(data, bins=10, color='skyblue', edgecolor='Blue')
plt.title("Histogram of Random Integers (1–100)")
plt.xlabel("Value Range")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--')
plt.show()
```

* np.random.randint(1, 101, 1000) generates 1000 random integers between 1 and 100.

* plt.hist() creates the histogram.

* bins=10 divides the range into 10 equal intervals.

* edgecolor='Blue' outlines the bars.

* The grid makes frequencies easier to compare.
